## Ejercicio 1 (3 puntos)

Implementa una función `factorial(n)` que dado un entero no negativo `n`, devuelva `n!`.

Recuerda que `0! = 1` y `n! = n * (n-1)!`.


```python
def factorial(n):
    ### BEGIN SOLUTION
    if n < 0:
        raise ValueError("n debe ser no negativo")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
    ### END SOLUTION
```

```python
# Tests visibles (los alumnos los ven)
assert factorial(0) == 1
assert factorial(1) == 1
assert factorial(5) == 120
```

```python
# Hidden tests — los alumnos NO los ven en su versión
### BEGIN HIDDEN TESTS
assert factorial(10) == 3628800
assert factorial(7) == 5040
try:
    factorial(-1)
    assert False, 'Debió lanzar ValueError'
except ValueError:
    pass
### END HIDDEN TESTS
```

## Ejercicio 2 (2 puntos)

Implementa `es_primo(n)` que devuelve `True` si `n` es un número primo, `False` en caso contrario.

Considera que los números menores que 2 NO son primos.

```python
def es_primo(n):
    ### BEGIN SOLUTION
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
    ### END SOLUTION
```

```python
assert es_primo(2) == True
assert es_primo(7) == True
assert es_primo(1) == False
assert es_primo(15) == False
### BEGIN HIDDEN TESTS
assert es_primo(0) == False
assert es_primo(-5) == False
assert es_primo(97) == True
assert es_primo(100) == False
### END HIDDEN TESTS
```
